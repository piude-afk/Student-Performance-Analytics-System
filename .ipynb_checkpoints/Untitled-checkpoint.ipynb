```python

# Keep the original data safe
marks_original = marks.copy()

# Remove duplicate rows
departments = departments.drop_duplicates()
students = students.drop_duplicates()
subjects = subjects.drop_duplicates()
marks = marks.drop_duplicates()

# Convert marks to numeric values
marks["marks"] = pd.to_numeric(marks["marks"], errors="coerce")

print("Data cleaning started.")
print("Missing marks:", marks["marks"].isna().sum())
```

```python

subject_means = marks.groupby("subject_id")["marks"].transform("mean")

marks["marks"] = marks["marks"].fillna(subject_means)

print("Missing marks after cleaning:", marks["marks"].isna().sum())
```

```python

print("Departments:", departments.shape)
print("Students:", students.shape)
print("Subjects:", subjects.shape)
print("Marks:", marks.shape)

display(marks.head())
```

```python

subject_stats = []

for subject_id, group in marks.groupby("subject_id"):
    values = group["marks"].to_numpy()

    subject_name = subjects.loc[
        subjects["subject_id"] == subject_id,
        "subject_name"
    ].iloc[0]

    subject_stats.append({
        "subject_name": subject_name,
        "mean": np.mean(values),
        "median": np.median(values),
        "standard_deviation": np.std(values)
    })

subject_stats_df = pd.DataFrame(subject_stats)

subject_stats_df[["mean", "median", "standard_deviation"]] = (
    subject_stats_df[["mean", "median", "standard_deviation"]].round(2)
)

display(subject_stats_df)
```

```python

student_subject_pivot = marks.pivot_table(
    index="student_id",
    columns="subject_id",
    values="marks",
    aggfunc="mean"
)

student_subject_pivot = student_subject_pivot.reindex(
    columns=subjects["subject_id"]
)

student_subject_pivot.columns = subjects.set_index(
    "subject_id"
).loc[student_subject_pivot.columns, "subject_name"].values

display(student_subject_pivot.head())
```

```python

student_results = student_subject_pivot.copy()

student_results["total_marks"] = student_results.sum(axis=1)

student_results["percentage"] = (
    student_results["total_marks"] / (len(subjects) * 100) * 100
).round(2)

display(student_results.head())
```

```python

conditions = [
    student_results["percentage"] >= 90,
    student_results["percentage"] >= 80,
    student_results["percentage"] >= 70,
    student_results["percentage"] >= 60,
    student_results["percentage"] >= 40
]

grades = ["A+", "A", "B", "C", "D"]

student_results["grade"] = np.select(
    conditions,
    grades,
    default="F"
)

display(student_results[["total_marks", "percentage", "grade"]].head())
```

```python

subject_columns = student_subject_pivot.columns

student_results["result"] = np.where(
    (student_results[subject_columns] >= 40).all(axis=1),
    "PASS",
    "FAIL"
)

display(student_results[["total_marks", "percentage", "grade", "result"]].head())
```

```python

student_results["rank"] = (
    student_results["percentage"]
    .rank(method="min", ascending=False)
    .astype(int)
)

display(
    student_results[
        ["total_marks", "percentage", "grade", "result", "rank"]
    ].sort_values("rank").head(10)
)
```

```python

student_results = student_results.reset_index()

student_results = pd.merge(
    student_results,
    students[["student_id", "student_name", "department_id"]],
    on="student_id",
    how="left"
)

student_results = pd.merge(
    student_results,
    departments[["department_id", "department_name"]],
    on="department_id",
    how="left"
)

top_5_students = student_results.sort_values(
    ["percentage", "student_id"],
    ascending=[False, True]
).head(5)

display(
    top_5_students[
        ["rank", "student_id", "student_name",
         "department_name", "total_marks", "percentage", "grade"]
    ]
)
```

```python

pass_fail_counts = student_results["result"].value_counts()

print("Pass/Fail count:")
display(pass_fail_counts)

print("Total students:", len(student_results))
```

```python

analysis_data = pd.merge(marks, students, on="student_id", how="inner")

analysis_data = pd.merge(
    analysis_data,
    departments,
    on="department_id",
    how="inner"
)

analysis_data = pd.merge(
    analysis_data,
    subjects,
    on="subject_id",
    how="inner"
)

department_averages = (
    analysis_data.groupby("department_name")["marks"]
    .mean()
    .round(2)
    .reset_index(name="department_average")
)

display(department_averages)
```

```python

student_results.to_csv("student_report.csv", index=False)
subject_stats_df.to_csv("subject_statistics.csv", index=False)
department_averages.to_csv("department_averages.csv", index=False)

print("Day 4 analysis results saved successfully!")
```

```python
# Calculate average marks for each subject
subject_avg = final_data.groupby("subject_name")["marks"].mean()

# Create bar chart
plt.figure(figsize=(10, 6))

bars = plt.bar(subject_avg.index, subject_avg.values)

# Add value labels
for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        height,
        f"{height:.2f}",
        ha="center",
        va="bottom"
    )

plt.title("Average Marks per Subject")
plt.xlabel("Subject Name")
plt.ylabel("Average Marks")
plt.xticks(rotation=45)
plt.tight_layout()

# Save chart
plt.savefig("average_marks_per_subject.png", dpi=300)

plt.show()
```

```python
# Create one row per student
pivot_table = final_data.pivot_table(
    index=["student_id", "student_name", "department_name"],
    columns="subject_name",
    values="marks",
    aggfunc="mean"
).reset_index()

# Identify subject columns
subject_columns = [
    col for col in pivot_table.columns
    if col not in ["student_id", "student_name", "department_name"]
]

# Calculate total and percentage
pivot_table["total"] = pivot_table[subject_columns].sum(axis=1)

pivot_table["percentage"] = (
    pivot_table["total"] / (len(subject_columns) * 100)
) * 100

# Assign grades
conditions = [
    pivot_table["percentage"] >= 90,
    pivot_table["percentage"] >= 80,
    pivot_table["percentage"] >= 70,
    pivot_table["percentage"] >= 60,
    pivot_table["percentage"] >= 40
]

grades = ["A+", "A", "B", "C", "D"]

pivot_table["grade"] = np.select(
    conditions,
    grades,
    default="F"
)

display(pivot_table.head())
```

```python
# Count students in each grade
grade_counts = pivot_table["grade"].value_counts().sort_index()

# Create pie chart
plt.figure(figsize=(8, 8))

plt.pie(
    grade_counts.values,
    labels=grade_counts.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Grade Distribution of Students")
plt.xlabel("Grade Categories")
plt.ylabel("Student Distribution")
plt.tight_layout()

# Save chart
plt.savefig("grade_distribution.png", dpi=300)

plt.show()
```

```python
# Remove missing marks
all_marks = final_data["marks"].dropna()

# Create histogram
plt.figure(figsize=(10, 6))

plt.hist(
    all_marks,
    bins=10,
    edgecolor="black"
)

# Add vertical pass-mark line
plt.axvline(
    x=40,
    linestyle="--",
    linewidth=2,
    label="Pass Mark = 40"
)

plt.title("Distribution of Marks")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.legend()
plt.tight_layout()

# Save chart
plt.savefig("marks_distribution_histogram.png", dpi=300)

plt.show()
```

```python
# Get department names
department_names = pivot_table["department_name"].dropna().unique()

# Prepare percentage data for each department
percentage_data = [
    pivot_table.loc[
        pivot_table["department_name"] == department,
        "percentage"
    ].dropna()
    for department in department_names
]

# Create box plot
plt.figure(figsize=(10, 6))

plt.boxplot(
    percentage_data,
    tick_labels=department_names
)

plt.title("Student Percentage by Department")
plt.xlabel("Department")
plt.ylabel("Percentage (%)")
plt.xticks(rotation=45)
plt.tight_layout()

# Save chart
plt.savefig("percentage_by_department_boxplot.png", dpi=300)

plt.show()
```
